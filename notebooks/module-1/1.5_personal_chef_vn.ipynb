```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from ipywidgets import FileUpload

uploader = FileUpload(
    accept=".jpg, .heic",
    multiple=True
)

display(uploader)

```

# Add Tools

```python
# Tools
from langchain.tools import tool
from typing import Dict, Any 
from tavily import TavilyClient

tavily_client = TavilyClient()

tools = []

@tool
def search(query: str) -> Dict[str, Any]:
    """Searches web for info"""
    return tavily_client.search(query)

tools.append(search)
```

```python

from langchain.agents import create_agent
agent = create_agent(model="openai:gpt-5-nano", tools=tools)
```

```python
import base64
uploaded_file = uploader.value[0]
content_mv = uploaded_file["content"]

# memory view to bytes
img_bytes = bytes(content_mv)

# decode as base64
img_b64 = base64.b64encode(img_bytes). decode("utf-8")
```

```python
from langchain.messages import HumanMessage, SystemMessage
system_message = SystemMessage(content='''
    You are a personal chef. 
    Given this image of items in my fridge, suggest recipes 
    I can make with the ingredients displayed.
''')

question = HumanMessage(content=[
    {"type": "image", "base64": img_b64, "mime_type": "image/jpg"}
])
```

```python
response = agent.invoke(
    {"messages": [system_message, question]}
)
```
